```python
import pandas as pd
import random
```

```python
education=['Graduate','Postgraduate']
discipline=['Computer Science','Statistics','Data Science','Computer Applications','Mathematics']
role=['Data Analyst Intern','Python Intern','Data Science Intern','Power BI Intern']
source=['LinkedIn','Referral','Instagram','Website','Naukri','Indeed']
n = 150
short_listed = []
assessment = []
interview = []
selection = []
offer = []
joining = []
for i in range(n):
    s = random.choice(['Yes', 'No'])
    short_listed.append(s)
    if s == "Yes":
        a = random.choices(['Completed', 'Pending'],weights=[75, 25])[0]
    else:
        a = "Rejected"
    assessment.append(a)
    if a == "Completed":
        iv = random.choices(['Selected', 'Pending'],weights=[75, 25])[0]
    elif a == "Pending":
        iv = "Pending"
    else:
        iv = "Rejected"
    interview.append(iv)
    if iv == "Selected":
        se = random.choices(['Selected', 'Pending'],weights=[75, 25])[0]
    elif iv == "Pending":
        se = "Pending"
    else:
        se = "Rejected"
    selection.append(se)
    if se == "Selected":
        o = random.choices(['Selected', 'Pending'],weights=[75, 25])[0]
    elif se == "Pending":
        o = "Pending"
    else:
        o = "Rejected"
    offer.append(o)
    if o == "Selected":
        j = random.choice(['Joined', 'Pending'])
    elif o == "Pending":
        j = "Pending"
    else:
        j = "Rejected"
    joining.append(j)
```

```python
data={
    'Application_ID':[f'APP{i}' for i in range(1,151)],
    'Candidate_ID':[f'C{i}' for i in range(1,151)],
    'Education':[random.choice(education) for i in range(150)],
    'Discipline':[random.choice(discipline) for i in range(150)],
    'Role':[random.choice(role) for i in range(150)],
    'Source':[random.choice(source) for i in range(150)],
    'Short_listed':short_listed,
    'Assessment': assessment,
    'Interview':interview,
    'Selection':selection,
    'Offer':offer,
    'Joining':joining
}
df=pd.DataFrame(data)
```
